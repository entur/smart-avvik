```python
from google.cloud import bigquery
import seaborn as sns
import matplotlib.pyplot as plt

client = bigquery.Client(project="ent-kixavvik-dev")

df = client.query("""
  SELECT
    category,
    ABS(predicted_duration_minutes - duration_minutes) AS diff
  FROM `disruptions.predictions`
""").to_dataframe()

sns.boxplot(data=df, x="category", y="diff")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

# sort categories by median so the eye has a natural order
order = df.groupby("category")["diff"].median().sort_values().index

sns.boxplot(data=df, x="category", y="diff",
            order=order, showfliers=False, ax=ax)

ax.set_xticklabels(ax.get_xticklabels(), rotation=40, ha="right")
ax.set_ylabel("|predicted − actual| (min)")
ax.set_xlabel("")
plt.tight_layout()
plt.show()
```

```python
df = client.query("""
  SELECT
    category,
    (predicted_duration_minutes - duration_minutes) / duration_minutes * 100 AS pct_error
  FROM `disruptions.predictions`
  WHERE duration_minutes > 0
""").to_dataframe()

fig, ax = plt.subplots(figsize=(12, 5))

order = df.groupby("category")["pct_error"].median().sort_values().index

sns.boxplot(data=df, x="category", y="pct_error",
            order=order, showfliers=False, ax=ax)

ax.axhline(0, color="grey", linewidth=1, linestyle="--")  # perfect-prediction line
ax.set_xticklabels(ax.get_xticklabels(), rotation=40, ha="right")
ax.set_ylabel("Prediction error (%)")
ax.set_xlabel("")
plt.tight_layout()
plt.show()
```

```python
df = client.query("""
  SELECT
    category,
    duration_minutes,
    predicted_duration_minutes
  FROM `disruptions.predictions`
""").to_dataframe()

actual_avg = df.groupby("category")["duration_minutes"].mean()
order = df.groupby("category")["predicted_duration_minutes"].median().sort_values().index

fig, ax = plt.subplots(figsize=(12, 5))

# Predicted: box at p25/median/p75 only — whiskers collapsed
sns.boxplot(data=df, x="category", y="predicted_duration_minutes",
            order=order, whis=(25, 75), showfliers=False, ax=ax,
            color="steelblue")

# Actual average per category, overlaid
ax.scatter(range(len(order)), [actual_avg[c] for c in order],
           color="crimson", s=60, zorder=10, label="Avg. actual duration")

ax.set_xticklabels(ax.get_xticklabels(), rotation=40, ha="right")
ax.set_ylabel("Duration (min)")
ax.set_xlabel("")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
df = client.query("""
  SELECT
    category,
    duration_minutes,
    predicted_duration_minutes
  FROM `disruptions.predictions`
""").to_dataframe()

fig, ax = plt.subplots(figsize=(9, 8))

sns.scatterplot(data=df, x="duration_minutes", y="predicted_duration_minutes",
                hue="category", alpha=0.5, s=20, palette="tab20", ax=ax)

# y = x reference line ("perfect prediction")
lim = max(df["duration_minutes"].max(), df["predicted_duration_minutes"].max())
ax.plot([0, lim], [0, lim], color="grey", linestyle="--", linewidth=1, label="Perfect prediction")

ax.set_xlabel("Actual duration (min)")
ax.set_ylabel("Predicted duration (min)")
ax.set_aspect("equal")
ax.legend(bbox_to_anchor=(1.02, 1), loc="upper left", fontsize=8)
plt.tight_layout()
plt.show()
```

```python
import plotly.express as px
import plotly.graph_objects as go

df = client.query("""
  SELECT
    category,
    duration_minutes,
    predicted_duration_minutes
  FROM `disruptions.predictions5nsb`
""").to_dataframe()

fig = px.scatter(
    df,
    x="duration_minutes",
    y="predicted_duration_minutes",
    color="category",
    opacity=0.5,
    color_discrete_sequence=px.colors.qualitative.Light24,  # ~tab20 equivalent
    labels={
        "duration_minutes": "Actual duration (min)",
        "predicted_duration_minutes": "Predicted duration (min)",
    },
)
fig.update_traces(marker=dict(size=5))

# y = x reference line ("perfect prediction")
lim = max(df["duration_minutes"].max(), df["predicted_duration_minutes"].max())
fig.add_trace(go.Scatter(
    x=[0, lim], y=[0, lim],
    mode="lines",
    line=dict(color="grey", dash="dash", width=1),
    name="Perfect prediction",
))

fig.update_layout(
    width=800, height=750,
    yaxis=dict(scaleanchor="x", scaleratio=1),   # equivalent to set_aspect("equal")
    legend=dict(x=1.02, y=1, xanchor="left", yanchor="top", font=dict(size=10)),
)

fig.show()
```
